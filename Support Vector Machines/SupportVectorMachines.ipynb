Support Vector Machines Exercise-1

```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.svm import SVC
from sklearn.model_selection import train_test_split
from sklearn import metrics

# Generate random dataset following a quadratic distribution
np.random.seed(42)

X = np.random.uniform(-5, 5, (200, 2))

# Quadratic decision boundary: y = x^2
y = (X[:, 1] > X[:, 0] ** 2 / 3).astype(int)

# Split the dataset into training and testing data
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.3, random_state=5
)

print("Training Data Shape:", X_train.shape)
print("Testing Data Shape:", X_test.shape)

# Create SVM model using RBF kernel
model = SVC(kernel='rbf', random_state=0)

# Train the model
model.fit(X_train, y_train)

# Predict the test data
y_pred = model.predict(X_test)

print("\nPredicted Values:")
print(y_pred)

# Confusion Matrix
conf_mat = metrics.confusion_matrix(y_test, y_pred)

print("\nConfusion Matrix:")
print(conf_mat)

# Accuracy
accuracy_score = metrics.accuracy_score(y_test, y_pred)

print("\nAccuracy Score:", accuracy_score)
print("Accuracy in Percentage:", int(accuracy_score * 100), "%")

# Classification Report
print("\nClassification Report:")
print(metrics.classification_report(y_test, y_pred))

# Visualize the dataset
plt.figure(figsize=(8, 6))

plt.scatter(
    X[y == 0, 0],
    X[y == 0, 1],
    label='Class 0'
)

plt.scatter(
    X[y == 1, 0],
    X[y == 1, 1],
    label='Class 1'
)

# Plot quadratic boundary
x_values = np.linspace(-5, 5, 400)
y_values = x_values ** 2 / 3

plt.plot(x_values, y_values, label='Quadratic Boundary')

plt.xlabel('X1')
plt.ylabel('X2')
plt.title('SVM with Quadratic Distribution')
plt.legend()
plt.grid(True)
plt.show()
```
